
# KELT-20 Gaia astrometry simulation

This notebook simulates Gaia-like astrometric data for a low-mass and high-mass solution to KELT-20

```python
# %% [markdown]
# Kepler helpers (for fitting) + multi-planet astrometric simulator using jaxoplanet

# ----------------- Orbit / astrometry helpers for FITTING ONLY -----------------
def kepler_E(M, e, tol=1e-12, maxiter=50):
    """Solve Kepler's equation M = E - e sin E for E."""
    E = M + (e*np.sin(M)) / (1 - np.sin(M + e) + np.sin(M))  # good initial guess
    for _ in range(maxiter):
        dE = -(E - e*np.sin(E) - M) / (1 - e*np.cos(E))
        E += dE
        if np.all(np.abs(dE) < tol):
            break
    return E

def orbital_basis(t, P, e, M0):
    """
    Return X,Y basis functions for a Keplerian ellipse at times t (years).
    X,Y are the standard elliptical coordinates in the orbital plane.
    """
    n = 2*np.pi / P
    M = n*t + M0
    E = kepler_E(M % (2*np.pi), e)
    X = np.cos(E) - e
    Y = np.sqrt(1 - e**2) * np.sin(E)
    return X, Y

def a_star_mas(P_yr, Mp_Mjup, Mstar_Msun=1.0, dist_pc=20.0):
    """Approximate astrometric semi-major axis of the star (mas) for sanity checks."""
    Mp_over_Ms = Mp_Mjup * MJUP_TO_MSUN / Mstar_Msun
    a_planet_AU = (P_yr**2 * Mstar_Msun)**(1/3)
    alpha_arcsec = Mp_over_Ms * a_planet_AU / dist_pc
    return alpha_arcsec * 1000.0  # mas

# --- jaxoplanet-backed MULTI-planet simulator (R_sun -> AU -> arcsec -> mas) ---
import jax
import jax.numpy as jnp
from jaxoplanet.orbits.keplerian import Central, System

def simulate_gaia_astrometry_multi(
    t, psi,
    planets,            # list of dicts: [{"P":yr,"e":...,"i":rad,"Omega":rad,"omega":rad,"M0":rad,"Mp":Mjup}, ...]
    Mstar=1.0, dist_pc=100.0,
    x0_true=0.0, y0_true=0.0, sigma_AL=0.01, seed=42
):
    """
    Simulate Gaia-like astrometry for a multi-planet system.

    Parameters
    ----------
    t : array (years, relative to mid-baseline)
    psi : array (radians), along-scan angle at each epoch
    planets : list of dict
        Each dict: {"a": au, "e":, "i": rad, "Omega": rad, "omega": rad, "M0": rad, "Mp": M_Jup}
    Mstar : float
        Stellar mass in M_sun.
    dist_pc : float
        Distance in parsec.
    sigma_AL : float
        Along-scan 1σ error [mas].

    Returns
    -------
    data : dict
        Contains t, psi, x_true, y_true, al_obs, al_true, sigma, and 'truth'.
    """
    t_days = jnp.asarray(t * 365.25)
    star   = Central(mass=jnp.asarray(Mstar), radius=jnp.asarray(1.0))
    sys    = System(star)

    # Build bodies
    for p in planets:
        n = 2.0 * jnp.pi / P_days
        t_peri_days = - p["M0"] / n
        sys = sys.add_body(
            semimajor=jnp.asarray(p["a"]),
            eccentricity=jnp.asarray(p["e"]),
            inclination=jnp.asarray(p["i"]),
            omega_peri=jnp.asarray(p["omega"]),
            asc_node=jnp.asarray(p["Omega"]),
            time_peri=t_peri_days,
            mass=jnp.asarray(p["Mp"] * MJUP_TO_MSUN),
        )

    bodies = sys.bodies

    # Sum the star's reflex due to each body (each central_position is in R_sun)
    x_sum = 0.0
    y_sum = 0.0
    for b in bodies:
        x_rsun, y_rsun, _ = b.central_position(t_days)
        x_sum = x_sum + x_rsun
        y_sum = y_sum + y_rsun

    # Convert R_sun -> AU -> arcsec (AU/d_pc) -> mas
    x_true = np.asarray(x_sum) * RSUN_TO_AU * (1000.0 / dist_pc) + x0_true
    y_true = np.asarray(y_sum) * RSUN_TO_AU * (1000.0 / dist_pc) + y0_true

    # Project to Gaia along-scan and add noise
    rng = np.random.default_rng(seed)
    al_true = x_true*np.cos(psi) + y_true*np.sin(psi)
    al_obs  = al_true + rng.normal(0, sigma_AL, size=len(al_true))

    return dict(
        t=t, psi=psi,
        x_true=x_true, y_true=y_true,
        al_obs=al_obs, al_true=al_true,
        sigma=sigma_AL,
        truth={"planets": planets},
    )


def make_ra_dec_observations(data):
    """
    Construct x_obs, y_obs that are perfectly consistent with the existing
    along-scan observations (al_obs) and the true AL model (al_true),
    and compute per-epoch RA/Dec uncertainties projected from AL sigma.

    This function REQUIRES data["al_obs"] and data["al_true"].
    It will raise a KeyError if they are missing.

    Notes
    -----
    Ensures that:
        x_obs*cos(psi) + y_obs*sin(psi) == al_obs
    exactly for every epoch.
    """

    # --- Require AL data ---
    if "al_obs" not in data or "al_true" not in data:
        raise KeyError(
            "make_ra_dec_observations requires data['al_obs'] and data['al_true']. "
            "These must be produced by simulate_gaia_astrometry_multi."
        )

    # --- Build x_obs, y_obs only if not already present ---
    if ("x_obs" not in data) or ("y_obs" not in data):
        eps = data["al_obs"] - data["al_true"]    # actual AL noise
        c, s = np.cos(data["psi"]), np.sin(data["psi"])

        x_obs = data["x_true"] + eps * c
        y_obs = data["y_true"] + eps * s

        data = dict(data, x_obs=x_obs, y_obs=y_obs)

    # --- Marginal RA/Dec uncertainties (for plotting only) ---
    sigma_ra  = data["sigma"] * np.abs(np.cos(data["psi"]))
    sigma_dec = data["sigma"] * np.abs(np.sin(data["psi"]))

    return data, sigma_ra, sigma_dec




```

```python
import numpy as np
import os
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator


# ---------------------------------------------------------------------
# Helpers
# ---------------------------------------------------------------------


def mutual_inclination_deg(i1, i2, Omega1, Omega2):
    """Compute mutual inclination (deg) from inclinations & nodes (rad)."""
    cos_imut = (
        np.cos(i1) * np.cos(i2)
        + np.sin(i1) * np.sin(i2) * np.cos(Omega1 - Omega2)
    )
    cos_imut = np.clip(cos_imut, -1.0, 1.0)
    return np.degrees(np.arccos(cos_imut))


def build_scan_times(duration_yr, ra_deg=20.0, dec_deg=-45.0):
    """
    Build Gaia scan times and along-scan angles over a given duration.

    Returns:
        t       : times relative to mid-baseline (for numerics)
        psi     : along-scan angles [rad]
        times_year : absolute times in years (this is what we want on the x-axis)
        T_span  : actual span in years
    """
    tstart = gaiascanlaw.tstart
    tend   = tstart + duration_yr
    times_year, scan_angles = gaiascanlaw.scanlaw(
        ra_deg, dec_deg, tstart=tstart, tend=tend
    )
    times_year = np.asarray(times_year)
    psi        = np.asarray(scan_angles)  # radians

    # Re-center time for numerical stability
    t_ref = np.mean(times_year)
    t     = times_year - t_ref   # years relative to mid-baseline

    T_span = times_year.max() - times_year.min()  # actual span
    return t, psi, times_year, T_span


def make_output_paths(duration_yr, imut_target_deg):
    """
    Decide folder and filename tags based on baseline & mutual inclination.
    Returns (folder_base, baseline_tag, geom_tag).
    """
    # Baseline tag
    if duration_yr <= 6.0:
        baseline_tag = "dr4"
        base_folder  = "gaia_dr4"
    else:
        baseline_tag = "dr5"
        base_folder  = "gaia_dr5"

    # Geometry tag
    if imut_target_deg == 0.0:
        geom_tag = "coplanar"
    else:
        geom_tag = "inclined"

    folder_base = "figures"
    return folder_base, baseline_tag, geom_tag


# ---------------------------------------------------------------------
# Plotting
# ---------------------------------------------------------------------

def plot_and_save_figures(
    data_2to1,
    data_eccentric,
    x1_aligned,
    y1_aligned,
    al_2_true,
    al_1_shifted,
    sigma_ra,
    sigma_dec,
    sigma_arr,
    folder_base,
    baseline_tag,
    geom_tag,
    color1,
    color2,
    t_plot,
    fig_suptitle
):
    """
    Make the three figures (sky-plane, RA/Dec vs time, AL vs time+residuals)
    and save them into the appropriate folder.

    t_plot: the times in years (times_year) for the x-axis.
    """
    # Common styles
    err_kw = dict(
        fmt='none',
        ecolor='black',
        elinewidth=1,
        alpha=0.36,
        capsize=0,
    )
    pt_kw = dict(
        marker='o',
        ls='none',
        ms=3,
        color='black',
        alpha=0.9,
    )

    # Convenience aliases
    x_obs   = data_2to1["x_obs"]
    y_obs   = data_2to1["y_obs"]
    x2_true = data_2to1["x_true"]
    y2_true = data_2to1["y_true"]
    t_rel   = data_2to1["t"]        # relative times used internally by the simulator
    al_obs  = data_2to1["al_obs"]
    psi_arr = data_2to1["psi"]

    # ---------------------------------------------------------------------
    # Plot 1: sky-plane (ΔRA*, ΔDec)
    # ---------------------------------------------------------------------
    plt.figure(figsize=(9, 9))

    for i in range(len(data_2to1["t"])):
        dx_err = data_2to1["sigma"] * np.cos(data_2to1["psi"][i])
        dy_err = data_2to1["sigma"] * np.sin(data_2to1["psi"][i])
        plt.plot(
            [x_obs[i] - dx_err, x_obs[i] + dx_err],
            [y_obs[i] - dy_err, y_obs[i] + dy_err],
            color='black', alpha=0.25, lw=1
        )

    plt.plot(x_obs, y_obs, 'o', ms=3, alpha=0.9,
             color='black', label=r"2:1 \textit{Gaia}-like data")

    plt.plot(x2_true, y2_true, color=color1, ls='-', label="2:1 model",
             lw=3, alpha=0.9, zorder=-100)
    plt.plot(x1_aligned, y1_aligned, color=color2, ls='-', label="eccentric model",
             lw=3, alpha=0.9, zorder=-100)

    plt.xlabel(r"$\Delta\alpha\,\cos\,\delta$ [mas]", fontsize=23)
    plt.ylabel(r"$\Delta\,\delta$ [mas]", fontsize=23)
    plt.legend(fontsize=23, loc=2)
    plt.gca().set_aspect("equal", adjustable="datalim")
    plt.grid(alpha=0.3)
    plt.tight_layout()

    skyplane_path = os.path.join(
        folder_base, f"ra_dec_{baseline_tag}_{geom_tag}.pdf"
    )
    plt.savefig(skyplane_path)
    plt.close()

    # ---------------------------------------------------------------------
    # Plot 2: RA* and Dec vs time + combined residuals
    # ---------------------------------------------------------------------
    fig, axes = plt.subplots(
        4, 1, figsize=(9, 9), sharex=True,
        gridspec_kw={"height_ratios": [4, 1.5, 4, 1.5]}
    )

    ax_ra, ax_ra_res, ax_dec, ax_dec_res = axes

    # vertical line location in absolute time coordinates
    dr4_time = np.min(t_plot) + 5.5  # adjust if you want a specific DR4 epoch

    # -------------------------
    # RA* vs time (main panel)
    # -------------------------
    if sigma_ra is not None:
        ax_ra.errorbar(t_plot, x_obs, yerr=sigma_ra, **err_kw)

    ax_ra.plot(t_plot, x_obs, label="2:1 Gaia-like data", **pt_kw)
    ax_ra.plot(t_plot, x2_true, color=color1, ls='-',
               label="2:1 model", lw=3, alpha=0.9, zorder=-100)
    ax_ra.plot(t_plot, x1_aligned, color=color2, ls='-',
               label="eccentric model", lw=3, alpha=0.9, zorder=-100)

    # vertical line & label in main panel only
    ax_ra.axvline(dr4_time, color="k", lw=2, ls="--", alpha=0.9)
    ax_ra.text(
        dr4_time, 0.3, r"\textit{Gaia} DR4 baseline",
        rotation=90,
        va="center",
        ha="right",
        fontsize=13,
        color="k",
        alpha=0.9,
        transform=ax_ra.get_xaxis_transform()
    )

    ax_ra.set_ylabel(r"$\Delta\alpha\,\cos\,\delta$ [mas]", fontsize=23)
    ax_ra.grid(alpha=0.3)

    # -------------------------
    # RA* residuals (both models in same subplot)
    # -------------------------
    res_ra_2to1 = x_obs - x2_true
    res_ra_ecc  = x_obs - x1_aligned

    ax_ra_res.axhline(0, color='k', lw=1, ls='-')
    ax_ra_res.axvline(dr4_time, color="k", lw=2, ls="--", alpha=0.9)

    if sigma_ra is not None:
        ax_ra_res.errorbar(
            t_plot, res_ra_2to1, yerr=sigma_ra,
            fmt='s', ms=3, alpha=0.7, color=color1,
            label="2:1 residual"
        )
        ax_ra_res.errorbar(
            t_plot, res_ra_ecc, yerr=sigma_ra,
            fmt='s', ms=3, alpha=0.7, color=color2,
            label="eccentric residual"
        )
    else:
        ax_ra_res.plot(
            t_plot, res_ra_2to1,
            's', ms=3, alpha=0.7, color=color1,
            label="2:1 residual"
        )
        ax_ra_res.plot(
            t_plot, res_ra_ecc,
            's', ms=3, alpha=0.7, color=color2,
            label="eccentric residual"
        )

    ax_ra_res.set_ylabel("Res [mas]", fontsize=18)
    ax_ra_res.grid(alpha=0.3)

    # -------------------------
    # Dec vs time (main panel)
    # -------------------------
    if sigma_dec is not None:
        ax_dec.errorbar(t_plot, y_obs, yerr=sigma_dec, **err_kw)

    ax_dec.plot(t_plot, y_obs, label="2:1 Gaia-like data", **pt_kw)
    ax_dec.plot(t_plot, y2_true, color=color1, ls='-',
                label="2:1 model", lw=3, alpha=0.9, zorder=-100)
    ax_dec.plot(t_plot, y1_aligned, color=color2, ls='-',
                label="eccentric model", lw=3, alpha=0.9, zorder=-100)

    # vertical line & label in main panel only
    ax_dec.axvline(dr4_time, color="k", lw=2, ls="--", alpha=0.9)
    ax_dec.text(
        dr4_time, 0.3, r"\textit{Gaia} DR4 baseline",
        rotation=90,
        va="center",
        ha="right",
        fontsize=13,
        color="k",
        alpha=0.9,
        transform=ax_dec.get_xaxis_transform()
    )

    ax_dec.set_ylabel(r"$\Delta\,\delta$ [mas]", fontsize=23)
    ax_dec.grid(alpha=0.3)

    # -------------------------
    # Dec residuals (both models in same subplot)
    # -------------------------
    res_dec_2to1 = y_obs - y2_true
    res_dec_ecc  = y_obs - y1_aligned

    ax_dec_res.axhline(0, color='k', lw=1, ls='-')
    ax_dec_res.axvline(dr4_time, color="k", lw=2, ls="--", alpha=0.9)

    if sigma_dec is not None:
        ax_dec_res.errorbar(
            t_plot, res_dec_2to1, yerr=sigma_dec,
            fmt='s', ms=3, alpha=0.7, color=color1,
            label="2:1 residual"
        )
        ax_dec_res.errorbar(
            t_plot, res_dec_ecc, yerr=sigma_dec,
            fmt='s', ms=3, alpha=0.7, color=color2,
            label="eccentric residual"
        )
    else:
        ax_dec_res.plot(
            t_plot, res_dec_2to1,
            's', ms=3, alpha=0.7, color=color1,
            label="2:1 residual"
        )
        ax_dec_res.plot(
            t_plot, res_dec_ecc,
            's', ms=3, alpha=0.7, color=color2,
            label="eccentric residual"
        )

    ax_dec_res.set_xlabel("Time [year]", fontsize=23)
    ax_dec_res.set_ylabel("Res [mas]", fontsize=18)
    ax_dec_res.grid(alpha=0.3)
    
    
    

    plt.tight_layout()

    radec_time_path = os.path.join(
        folder_base, f"ra_dec_time_{baseline_tag}_{geom_tag}.pdf"
    )
    plt.savefig(radec_time_path)
    plt.close()

    # ---------------------------------------------------------------------
    # Plot 3: Along-scan (AL) vs time + combined residuals
    # ---------------------------------------------------------------------
    fig, (ax_al, ax_al_res) = plt.subplots(
        2, 1, figsize=(9, 6), sharex=True,
        gridspec_kw={"height_ratios": [4, 1.5]}
    )

    # vertical line location in absolute time coordinates
    dr4_time = np.min(t_plot) + 5.5  # same convention as RA/Dec

    # Main AL panel
    if sigma_arr is not None:
        ax_al.errorbar(t_plot, al_obs, yerr=sigma_arr, **err_kw)

    ax_al.plot(
        t_plot, al_obs,
        label="2:1 Gaia-like data",
        **pt_kw
    )
    ax_al.plot(
        t_plot, al_2_true,
        color=color1, ls='-', label="2:1 model",
        alpha=0.9, lw=3, zorder=-100
    )
    ax_al.plot(
        t_plot, al_1_shifted,
        color=color2, ls='-', label="eccentric model",
        alpha=0.9, lw=3, zorder=-100
    )

    # DR4 line + label (same style as RA/Dec main panels)
    ax_al.axvline(dr4_time, color="k", lw=2, ls="--", alpha=0.9)
    ax_al.text(
        dr4_time, 0.3, r"\textit{Gaia} DR4 baseline",
        rotation=90,
        va="center",
        ha="right",
        fontsize=13,
        color="k",
        alpha=0.9,
        transform=ax_al.get_xaxis_transform()
    )

    ax_al.set_ylabel(r"$\Delta\eta$ [mas]", fontsize=23)
    ax_al.grid(alpha=0.3)

    # Residuals (both models in a single subplot)
    res_2to1 = al_obs - al_2_true
    res_ecc  = al_obs - al_1_shifted

    ax_al_res.axhline(0, color='k', lw=1, ls='-')
    ax_al_res.axvline(dr4_time, color="k", lw=2, ls="--", alpha=0.9)

    if sigma_arr is not None:
        ax_al_res.errorbar(
            t_plot, res_2to1, yerr=sigma_arr,
            fmt='s', ms=3, alpha=0.7, color=color1,
            label="2:1 residual"
        )
        ax_al_res.errorbar(
            t_plot, res_ecc, yerr=sigma_arr,
            fmt='s', ms=3, alpha=0.7, color=color2,
            label="eccentric residual"
        )
    else:
        ax_al_res.plot(
            t_plot, res_2to1,
            's', ms=3, alpha=0.7, color=color1,
            label="2:1 residual"
        )
        ax_al_res.plot(
            t_plot, res_ecc,
            's', ms=3, alpha=0.7, color=color2,
            label="eccentric residual"
        )

    ax_al_res.set_xlabel("Time [year]", fontsize=23)
    ax_al_res.set_ylabel("Res [mas]", fontsize=18)
    ax_al_res.grid(alpha=0.3)

    plt.tight_layout()

    al_path = os.path.join(
        folder_base, f"AL_{baseline_tag}_{geom_tag}.pdf"
    )
    plt.savefig(al_path)
    plt.close()

    
    
    # ---------------------------------------------------------------------
    # Plot 4 Combined Figure:
    # Col 0: sky-plane (ΔRA*, ΔDec)
    # Col 1: AL vs time + residuals
    # Col 2: RA vs time + residuals, Dec vs time + residuals
    # ---------------------------------------------------------------------
    fig = plt.figure(figsize=(18, 7))

    # Outer grid: 1 row × 3 columns
    outer_gs = fig.add_gridspec(
        1, 3,
        width_ratios=[1.1, 1.0, 1.0],
        wspace=0.3
    )

    # === LEFT: Sky-plane ===
    ax_left = fig.add_subplot(outer_gs[0, 0])

    for i in range(len(data_2to1["t"])):
        dx_err = data_2to1["sigma"] * np.cos(data_2to1["psi"][i])
        dy_err = data_2to1["sigma"] * np.sin(data_2to1["psi"][i])
        ax_left.plot(
            [x_obs[i] - dx_err, x_obs[i] + dx_err],
            [y_obs[i] - dy_err, y_obs[i] + dy_err],
            color='black', alpha=0.25, lw=1
        )

    ax_left.plot(
        x_obs, y_obs, 'o', ms=3, alpha=0.9,
        color='black', label=r"2:1 \textit{Gaia}-like data"
    )
    ax_left.plot(
        x2_true, y2_true, color=color1, ls='-', label="2:1 model",
        lw=3, alpha=0.9, zorder=-100
    )
    ax_left.plot(
        x1_aligned, y1_aligned, color=color2, ls='-', label="eccentric model",
        lw=3, alpha=0.9, zorder=-100
    )

    ax_left.set_xlabel(r"$\Delta\alpha\,\cos\,\delta$ [mas]", fontsize=23)
    ax_left.set_ylabel(r"$\Delta\,\delta$ [mas]", fontsize=23)
    ax_left.legend(fontsize=16, loc=2)
    ax_left.set_aspect("equal", adjustable="datalim")
    ax_left.grid(alpha=0.3)

    # vertical line location for time plots
    dr4_time = np.min(t_plot) + 5.5

    # === MIDDLE: AL vs time + residuals (own column) ===
    mid_gs = outer_gs[0, 1].subgridspec(
        2, 1,
        height_ratios=[4, 1.5],
        hspace=0.05
    )

    ax_al     = fig.add_subplot(mid_gs[0, 0])
    ax_al_res = fig.add_subplot(mid_gs[1, 0], sharex=ax_al)
    ax_al.tick_params(labelbottom=False)


    # AL main
    if sigma_arr is not None:
        ax_al.errorbar(t_plot, al_obs, yerr=sigma_arr, **err_kw)

    ax_al.plot(t_plot, al_obs, **pt_kw)
    ax_al.plot(t_plot, al_2_true, color=color1, ls='-', lw=3, alpha=0.7, zorder=-10)
    ax_al.plot(t_plot, al_1_shifted, color=color2, ls='-', lw=3, alpha=0.7, zorder=-10)

    ax_al.axvline(dr4_time, color="k", lw=2, ls="--", alpha=0.9)
    ax_al.text(
        dr4_time, 0.2, r"\textit{Gaia} DR4 baseline",
        rotation=90,
        va="center",
        ha="right",
        fontsize=12,
        color="k",
        transform=ax_al.get_xaxis_transform()
    )

    ax_al.set_ylabel(r"$\Delta\eta$ [mas]", fontsize=18)
    ax_al.grid(alpha=0.3)

    # AL residuals (single panel, both models)
    res_2to1 = al_obs - al_2_true
    res_ecc  = al_obs - al_1_shifted

    ax_al_res.axhline(0, color='k', lw=1)
    ax_al_res.axvline(dr4_time, color="k", lw=2, ls="--", alpha=0.9)
    

    ax_al_res.errorbar(
        t_plot, res_2to1, yerr=sigma_arr,
        fmt='s', ms=3, alpha=0.5, color=color1
    )
    ax_al_res.errorbar(
        t_plot, res_ecc, yerr=sigma_arr,
        fmt='s', ms=3, alpha=0.5, color=color2
    )
    ax_al_res.set_ylabel("Res", fontsize=16)
    ax_al_res.grid(alpha=0.3)

    # === RIGHT: RA & Dec vs time + residuals (stacked) ===
    right_gs = outer_gs[0, 2].subgridspec(
        4, 1,
        height_ratios=[4, 1.5, 4, 1.5],
        hspace=0.05
    )

    ax_ra      = fig.add_subplot(right_gs[0, 0], sharex=ax_al)
    ax_ra_res  = fig.add_subplot(right_gs[1, 0], sharex=ax_al)
    ax_dec     = fig.add_subplot(right_gs[2, 0], sharex=ax_al)
    ax_dec_res = fig.add_subplot(right_gs[3, 0], sharex=ax_al)

    # RA main
    if sigma_ra is not None:
        ax_ra.errorbar(t_plot, x_obs, yerr=sigma_ra, **err_kw)

    ax_ra.plot(t_plot, x_obs, **pt_kw)
    ax_ra.plot(t_plot, x2_true, color=color1, ls='-', lw=3, alpha=0.7, zorder=-10)
    ax_ra.plot(t_plot, x1_aligned, color=color2, ls='-', lw=3, alpha=0.7, zorder=-10)

    ax_ra.axvline(dr4_time, color="k", lw=2, ls="--", alpha=0.9)
    ax_ra.text(
        dr4_time, 0.4, r"\textit{Gaia} DR4 baseline",
        rotation=90,
        va="center",
        ha="right",
        fontsize=12,
        color="k",
        transform=ax_ra.get_xaxis_transform()
    )

    ax_ra.set_ylabel(r"$\Delta\alpha\,\cos\,\delta$ [mas]", fontsize=18)
    ax_ra.grid(alpha=0.3)

    # RA residuals
    res_ra_2to1 = x_obs - x2_true
    res_ra_ecc  = x_obs - x1_aligned

    ax_ra_res.axhline(0, color='k', lw=1)
    ax_ra_res.axvline(dr4_time, color="k", lw=2, ls="--", alpha=0.9)

    ax_ra_res.errorbar(
        t_plot, res_ra_2to1, yerr=sigma_ra,
        fmt='s', ms=3, alpha=0.5, color=color1
    )
    ax_ra_res.errorbar(
        t_plot, res_ra_ecc, yerr=sigma_ra,
        fmt='s', ms=3, alpha=0.5, color=color2
    )
    ax_ra_res.set_ylabel("Res", fontsize=16)
    ax_ra_res.grid(alpha=0.3)

    # Dec main
    if sigma_dec is not None:
        ax_dec.errorbar(t_plot, y_obs, yerr=sigma_dec, **err_kw)

    ax_dec.plot(t_plot, y_obs, **pt_kw)
    ax_dec.plot(t_plot, y2_true, color=color1, ls='-', lw=3, alpha=0.7, zorder=-10)
    ax_dec.plot(t_plot, y1_aligned, color=color2, ls='-', lw=3, alpha=0.7, zorder=-10)

    ax_dec.axvline(dr4_time, color="k", lw=2, ls="--", alpha=0.9)
    ax_dec.text(
        dr4_time, 0.4, r"\textit{Gaia} DR4 baseline",
        rotation=90,
        va="center",
        ha="right",
        fontsize=12,
        color="k",
        transform=ax_dec.get_xaxis_transform()
    )

    ax_dec.set_ylabel(r"$\Delta\,\delta$ [mas]", fontsize=18)
    ax_dec.grid(alpha=0.3)

    # Dec residuals
    res_dec_2to1 = y_obs - y2_true
    res_dec_ecc  = y_obs - y1_aligned

    ax_dec_res.axhline(0, color='k', lw=1)
    ax_dec_res.axvline(dr4_time, color="k", lw=2, ls="--", alpha=0.9)

    ax_dec_res.errorbar(
        t_plot, res_dec_2to1, yerr=sigma_dec,
        fmt='s', ms=3, alpha=0.5, color=color1
    )
    ax_dec_res.errorbar(
        t_plot, res_dec_ecc, yerr=sigma_dec,
        fmt='s', ms=3, alpha=0.5, color=color2
    )

    ax_dec_res.set_ylabel("Res", fontsize=16)
    ax_dec_res.set_xlabel("Time [year]", fontsize=18)
    ax_dec_res.grid(alpha=0.3)


    # Force integer ticks on the shared time axis
    for ax in [ax_al, ax_ra, ax_dec]:
        ax.xaxis.set_major_locator(MaxNLocator(integer=True))



    # --- FINAL title + layout + save ---
    fig.suptitle(fig_suptitle, fontsize=36)
    fig.tight_layout(rect=[0, 0, 1, 0.96])

    combined_path = os.path.join(
        folder_base, f"skyplane_and_radec_{baseline_tag}_{geom_tag}.pdf"
    )
    plt.savefig(combined_path)
    plt.close()





# ---------------------------------------------------------------------
# Driver
# ---------------------------------------------------------------------

def run_case(duration_yr, imut_target_deg, label, color1, color2, fig_suptitle):
    """
    Run one simulation+fit case, make plots, and print a LaTeX caption line.
    """
    # 1) Build Gaia sampling
    t, psi, times_year, T_span = build_scan_times(duration_yr)
    N_AL = len(t)

    # 2) Orbital geometry
    i1     = np.radians(i1_deg)
    Omega1 = np.radians(Omega1_deg)
    omega1 = np.radians(omega1_deg)
    M0_1   = np.radians(M0_1_deg)

    if imut_target_deg == 0.0:
        # Coplanar: i2=i1, Omega2=Omega1
        i2     = i1
        Omega2 = Omega1
    else:
        # Impose |i1 - i2| ≈ imut_target_deg with same node
        i2     = np.radians(i1_deg - imut_target_deg)
        Omega2 = Omega1

    omega2 = omega1
    M0_2   = M0_1

    imut_actual = mutual_inclination_deg(i1, i2, Omega1, Omega2)

    # 3) Define planet dictionaries
    planets_2to1 = [
        dict(P=P1, e=e1, i=i1, Omega=Omega1, omega=omega1, M0=M0_1, Mp=Mp1),
        dict(P=P2, e=e2, i=i2, Omega=Omega2, omega=omega2, M0=M0_2, Mp=Mp2),
    ]
    planet_eccentric = [
        dict(P=P1, e=e_eff, i=i1, Omega=Omega1, omega=omega1, M0=M0_1, Mp=Mp1),
    ]

    # 4) Simulate Gaia astrometry
    data_2to1 = simulate_gaia_astrometry_multi(
        t, psi,
        planets=planets_2to1,
        Mstar=MSTAR,
        dist_pc=DIST_PC,
        sigma_AL=sigma_AL,
    )
    data_2to1, sigma_ra, sigma_dec = make_ra_dec_observations(data_2to1)

    data_eccentric = simulate_gaia_astrometry_multi(
        t, psi,
        planets=planet_eccentric,
        Mstar=MSTAR,
        dist_pc=DIST_PC,
        sigma_AL=sigma_AL,
    )

    # 5) residuals in AL space
    al_obs  = data_2to1["al_obs"]
    psi_arr = data_2to1["psi"]
    sigma_val = data_2to1["sigma"]

    if np.isscalar(sigma_val):
        sigma_arr = np.full_like(al_obs, sigma_val, dtype=float)
    else:
        sigma_arr = np.asarray(sigma_val, dtype=float)

    al_2_true = data_2to1["al_true"]      # 2:1 model (truth)
    al_1_true = data_eccentric["al_true"] # eccentric model


    # Align eccentric orbit with 2:1 in sky plane by fitting constant offsets
    x2_true, y2_true = data_2to1["x_true"], data_2to1["y_true"]
    x1_true, y1_true = data_eccentric["x_true"], data_eccentric["y_true"]

    dx_fit = np.mean(x2_true - x1_true)
    dy_fit = np.mean(y2_true - y1_true)

    x1_aligned = x1_true + dx_fit
    y1_aligned = y1_true + dy_fit

    al_1_shifted = x1_aligned * np.cos(psi_arr) + y1_aligned * np.sin(psi_arr)


    # 6) Figure saving
    folder_base, baseline_tag, geom_tag = make_output_paths(duration_yr, imut_target_deg)
    plot_and_save_figures(
        data_2to1,
        data_eccentric,
        x1_aligned,
        y1_aligned,
        al_2_true,
        al_1_shifted,
        sigma_ra,
        sigma_dec,
        sigma_arr,
        folder_base,
        baseline_tag,
        geom_tag,
        color1,
        color2,
        times_year,
        fig_suptitle
    )

    # 7) Build LaTeX caption line
    caption = r"""
\textit{{{label}}} ($T_\mathrm{{span}}={Tspan:.1f}\,\mathrm{{yr}}$, $i_\mathrm{{mut}}\approx{imut:.0f}^\circ$): 
2{{:}}1 system with $P_1={P1:.1f}\,\mathrm{{yr}}$, $M_{{p,1}}={Mp1:.0f}\,M_\mathrm{{Jup}}$, $e_1={e1:.1f}$, 
$P_2={P2:.1f}\,\mathrm{{yr}}$, $M_{{p,2}}={Mp2:.2f}\,M_\mathrm{{Jup}}$, $e_2={e2:.1f}$, 
compared to a single-planet model with $P={P1:.1f}\,\mathrm{{yr}}$, $e={eeff:.1f}$, $M_p={Mp1:.0f}\,M_\mathrm{{Jup}}$; 
$N_\mathrm{{AL}}={N_AL}$.
""".format(
        label=label,
        Tspan=duration_yr,
        imut=imut_actual,
        P1=P1,
        Mp1=Mp1,
        e1=e1,
        P2=P2,
        Mp2=Mp2,
        e2=e2,
        eeff=e_eff,
        N_AL=N_AL,
    ).strip()

    print("\n" + "-"*80)
    print(caption)
    print("-"*80 + "\n")

    return {
        "label": label,
        "duration_yr": duration_yr,
        "imut_target_deg": imut_target_deg,
        "imut_actual_deg": imut_actual,
        "N_AL": N_AL,
        "T_span": T_span,
        "caption": caption,
        "folder_base": folder_base,
        "baseline_tag": baseline_tag,
        "geom_tag": geom_tag,
    }
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt

# ---- Gaia scan law ----
try:
    import gaiascanlaw
except ImportError as e:
    raise ImportError("Please `pip install gaiascanlaw` before running this notebook.") from e

# ----------------- Global constants & setup -----------------
MJUP_TO_MSUN = 0.0009543          # 1 M_Jup in M_Sun
RSUN_TO_AU   = 0.00465047         # 1 R_sun in AU

MSTAR   = 1.89    # M_sun
DIST_PC = 136.98   # pc
sigma_AL = 0.054 # mas, along-scan noise



# Planet setup (same for all runs)
P1    = 5.2        # yr, outer planet
Mp1   = 12.0       # M_Jup, outer planet
e_eff = 0.3        # eccentricity of single-planet model
Mp2   = Mp1 * e_eff * 0.79  # M_Jup, inner planet

e1 = 0.0  # outer (2:1 case)
e2 = 0.0  # inner (2:1 case)

# Base geometry for the outer planet
i1_deg     = 60.0
Omega1_deg = 40.0
omega1_deg = 0.0
M0_1_deg   = 0.0



# ---------------------------------------------------------------------
# Run the four cases:
#   - DR4:  5.5 yr, i_mut = 0°
#   - DR4:  5.5 yr, i_mut = 45°
#   - DR5: 10.5 yr, i_mut = 0°
#   - DR5: 10.5 yr, i_mut = 45°
# ---------------------------------------------------------------------

results = []
results.append(run_case(duration_yr=5.5, imut_target_deg=0.0,  label="DR4, coplanar", color1='#1A85FF', color2='#D41159', fig_suptitle='DR4 coplanar'))
results.append(run_case(duration_yr=5.5, imut_target_deg=15.0, label=r"DR4, $i_\mathrm{mut}=15^\circ$", color1='#1A85FF', color2='#D41159', fig_suptitle='DR4 inclined'))
results.append(run_case(duration_yr=10.5, imut_target_deg=0.0,  label="DR5, coplanar", color1='#1A85FF', color2='#D41159', fig_suptitle=r'coplanar, $i_\mathrm{mut}=0^\circ$'))
results.append(run_case(duration_yr=10.5, imut_target_deg=15.0, label=r"DR5, $i_\mathrm{mut}=15^\circ$", color1='#1A85FF', color2='#D41159', fig_suptitle=r"mutually inclined, $i_\mathrm{mut}=15^\circ$"))


```
